# The Quest for x

This notebook turns a multi-level numerical linear algebra adventure into standalone HTML. The diagram begins in `quest_for_x.yaml`; Python only validates and renders it.

Install FlowYAML into the active notebook kernel once with `%pip install flowyaml`.

```python
from pathlib import Path

import flowyaml

source = Path('quest_for_x.yaml')
artifact = Path('quest_for_x.html')
print('FlowYAML:', flowyaml.__version__)
```

```python
issues = flowyaml.validate(source)
if issues:
    raise ValueError('\n'.join(str(issue) for issue in issues))

print('Validation: OK')
print('Navigable models:', flowyaml.models(source))
```

```python
result = flowyaml.write_html(
    source,
    artifact,
    model_id='quest_for_x',
    output='document',
    assets='inline',
    theme='dornelles_multitech',
    instance_id='quest-for-x',
)
print('Standalone artifact:', result.resolve())
```
